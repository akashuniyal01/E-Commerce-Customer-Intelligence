```python
# ============================================
# 1. IMPORT LIBRARIES
# ============================================

import pandas as pd
import numpy as np
import unicodedata
import re
from rapidfuzz import process, fuzz


# ============================================
# 2. NORMALIZE CITY NAMES
# ============================================

def normalize_city(city):

    if pd.isna(city):
        return np.nan

    city = str(city).strip().lower()

    # Remove accents
    city = unicodedata.normalize('NFKD', city)
    city = ''.join(
        c for c in city
        if not unicodedata.combining(c)
    )

    # Replace -, _, / with space
    city = re.sub(r'[-_/]', ' ', city)

    # Remove other punctuation
    city = re.sub(r'[^\w\s]', ' ', city)

    # Remove extra spaces
    city = re.sub(r'\s+', ' ', city).strip()

    return city


df['city_clean'] = df['geolocation_city'].apply(normalize_city)


# ============================================
# 3. CHECK REDUCTION
# ============================================

print("Original unique cities:",
      df['geolocation_city'].nunique())

print("After normalization:",
      df['city_clean'].nunique())


# ============================================
# 4. FIND DIFFERENT RAW VALUES THAT
#    BECAME THE SAME CITY
# ============================================

city_variations = (
    df.groupby('city_clean')['geolocation_city']
      .unique()
)

duplicates = city_variations[
    city_variations.apply(len) > 1
]

print("\nExamples of normalized duplicates:")
print(duplicates.head(30))


# ============================================
# 5. CREATE UNIQUE CITY + STATE LIST
# ============================================

city_state = (
    df[['city_clean', 'geolocation_state']]
    .dropna()
    .drop_duplicates()
)


# ============================================
# 6. FIND POSSIBLE SPELLING ERRORS
#    WITHIN THE SAME STATE
# ============================================

suggestions = []

for state, group in city_state.groupby('geolocation_state'):

    cities = group['city_clean'].unique()

    for city in cities:

        matches = process.extract(
            city,
            cities,
            scorer=fuzz.ratio,
            limit=3
        )

        for match, score, _ in matches:

            # Don't compare city with itself
            if city == match:
                continue

            # Only keep high similarity
            if score >= 90:

                suggestions.append({
                    'state': state,
                    'city': city,
                    'possible_match': match,
                    'similarity': score
                })


suggestions_df = pd.DataFrame(suggestions)


# ============================================
# 7. REMOVE DUPLICATE PAIRS
# ============================================

if not suggestions_df.empty:

    suggestions_df['pair'] = suggestions_df.apply(
        lambda x: tuple(sorted([
            x['city'],
            x['possible_match']
        ])),
        axis=1
    )

    suggestions_df = (
        suggestions_df
        .drop_duplicates(subset=['state', 'pair'])
        .drop(columns='pair')
        .sort_values(
            ['similarity'],
            ascending=False
        )
    )


# ============================================
# 8. DISPLAY POSSIBLE SPELLING ERRORS
# ============================================

print("\nPossible spelling errors:")

print(
    suggestions_df[
        suggestions_df['similarity'] >= 90
    ].head(100)
)


# ============================================
# 9. CREATE YOUR VERIFIED CORRECTION MAP
# ============================================

# IMPORTANT:
# Add ONLY the corrections that you manually verify.

city_mapping = {

    # Example:
    # 'sao paul': 'sao paulo',
    # 'campinass': 'campinas',
    # 'belo horiznte': 'belo horizonte'

}


# ============================================
# 10. APPLY VERIFIED CORRECTIONS
# ============================================

df['city_clean'] = (
    df['city_clean']
    .replace(city_mapping)
)


# ============================================
# 11. FINAL CHECK
# ============================================

print("\nFinal unique cities:",
      df['city_clean'].nunique())

print("\nTop cities:")
print(
    df['city_clean']
    .value_counts()
    .head(20)
)
```
